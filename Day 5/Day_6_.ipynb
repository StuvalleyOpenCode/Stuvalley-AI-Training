1. Zero-Shot Prompting (No Examples Given)

```python
import openai

response = openai.ChatCompletion.create(
    model="gpt-4",
    messages=[{"role": "user", "content": "What is the capital of Japan?"}]
)

print(response["choices"][0]["message"]["content"])
```

2. Few-Shot Prompting (Providing Examples)

```python
prompt = """
Translate the following words into Spanish:
- Apple → Manzana
- Car → Coche
- House → ?
"""

response = openai.ChatCompletion.create(
    model="gpt-4",
    messages=[{"role": "user", "content": prompt}]
)

print(response["choices"][0]["message"]["content"])
```

3. Chain-of-Thought (CoT) Prompting (Step-by-Step Reasoning)

```python
prompt = """
Lisa has 4 chocolates. She buys 3 more and gives 2 to her friend.
How many does she have now? Explain step by step.
"""

response = openai.ChatCompletion.create(
    model="gpt-4",
    messages=[{"role": "user", "content": prompt}]
)

print(response["choices"][0]["message"]["content"])
```

4. Role-Based Prompting (Assigning a Role)

```python
prompt = "You are a history professor. Explain the French Revolution in simple words."

response = openai.ChatCompletion.create(
    model="gpt-4",
    messages=[{"role": "user", "content": prompt}]
)

print(response["choices"][0]["message"]["content"])
```

5. Instruction-Based Prompting (Giving Clear Instructions)

```python
prompt = """
Summarize this paragraph in one sentence:
'The sun is the closest star to Earth. It provides light and heat, which are essential for life. Without the sun, the planet would be too cold for living beings.'
"""

response = openai.ChatCompletion.create(
    model="gpt-4",
    messages=[{"role": "user", "content": prompt}]
)

print(response["choices"][0]["message"]["content"])
```

6. Multimodal Prompting (Text + Image)

```python
import openai
import base64

# Load an image (replace with your image path)
with open("/content/dog.jpg", "rb") as img:
    img_base64 = base64.b64encode(img.read()).decode("utf-8")

response = openai.ChatCompletion.create(
    model="gpt-4-vision-preview",
    messages=[
        {"role": "user", "content": "Describe this image."},
        {"role": "user", "content": {"type": "image", "image": img_base64}}
    ]
)

print(response["choices"][0]["message"]["content"])
```

7. Self-Consistency Prompting (Multiple Answers & Best Choice)

```python
prompt = """
Rewrite this sentence to sound more professional:
'I wanna let you know about a cool new thing.'
"""

response = openai.ChatCompletion.create(
    model="gpt-4",
    messages=[{"role": "user", "content": prompt}]
)

print(response["choices"][0]["message"]["content"])
```
