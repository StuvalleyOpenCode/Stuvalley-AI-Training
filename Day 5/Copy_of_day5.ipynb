```python
from PyPDF2 import PdfReader
from langchain.embeddings.openai import OpenAIEmbeddings
from langchain.text_splitter import CharacterTextSplitter
from langchain.vectorstores import FAISS
```

```python
# provide the path of  pdf file/files.
pdfreader = PdfReader('/content/Gowtham intreme report.pdf')
```

```python
from typing_extensions import Concatenate
# read text from pdf
raw_text = ''
for i, page in enumerate(pdfreader.pages):
    content = page.extract_text()
    if content:
        raw_text += content
```

```python
raw_text
```

```python
# We need to split the text using Character Text Split such that it sshould not increse token size
text_splitter = CharacterTextSplitter(
    separator = "\n",
    chunk_size = 800,
    chunk_overlap  = 200,
    length_function = len,
)
texts = text_splitter.split_text(raw_text)
```

```python
len(texts)
```

```python
# Download embeddings from OpenAI
embeddings = OpenAIEmbeddings()
```

```python
document_search = FAISS.from_texts(texts, embeddings)
```

```python
document_search
```

```python
from langchain.chains.question_answering import load_qa_chain
from langchain.llms import OpenAI
```

```python
chain = load_qa_chain(OpenAI(), chain_type="stuff")
```

```python
query = "what is the name"
docs = document_search.similarity_search(query)
chain.run(input_documents=docs, question=query)
```

```python
query = "what he done in his projects"
docs = document_search.similarity_search(query)
chain.run(input_documents=docs, question=query)
```

```python
import re

# Function to replace all variations of 'session id' with '[REDACTED]'
def redact_session_id(message):
    # Updated pattern to match 'session id' in different formats
    pattern = r"session[\s_]*id"
    if re.search(pattern, message, re.IGNORECASE):
        message = re.sub(pattern, "[REDACTED]", message, flags=re.IGNORECASE)
    return message

# Test cases with different variations of 'session id'
test_messages = [
    "The session id is 12345.",        # with space
    "Please check the SessionId 67890.",  # camelCase
    "User's SessionID was invalid.",    # uppercase 'ID'
    "session_id needs to be updated.",  # underscore
    "This is session__id 11223.",       # double underscore
    "sessionID should not be logged."   # mixed camelCase
]

# Testing each message
for i, test in enumerate(test_messages, 1):
    redacted_message = redact_session_id(test)
    print(f"Test {i}: {redacted_message}")
```
