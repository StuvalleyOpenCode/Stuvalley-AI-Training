1️⃣ Word Tokenization
Using NLTK and spaCy

```python
import nltk
from nltk.tokenize import word_tokenize
import spacy

nltk.download('punkt')

text = "Machine learning is fun!"
# Using NLTK
tokens_nltk = word_tokenize(text)
print("Word Tokenization (NLTK):", tokens_nltk)

# Using spaCy
nlp = spacy.load("en_core_web_sm")
tokens_spacy = [token.text for token in nlp(text)]
print("Word Tokenization (spaCy):", tokens_spacy)
```

2️⃣ Sentence Tokenization
Using NLTK and spaCy

```python
from nltk.tokenize import sent_tokenize

text = "Hello! How are you? I am fine."

# Using NLTK
sentences_nltk = sent_tokenize(text)
print("Sentence Tokenization (NLTK):", sentences_nltk)

# Using spaCy
doc = nlp(text)
sentences_spacy = [sent.text for sent in doc.sents]
print("Sentence Tokenization (spaCy):", sentences_spacy)
```

3️⃣ Subword Tokenization (Byte Pair Encoding - BPE, WordPiece, Unigram)
Using Hugging Face’s Tokenizers

```python
from tokenizers import Tokenizer, models, pre_tokenizers, trainers

# Create a BPE tokenizer
tokenizer = Tokenizer(models.BPE())
tokenizer.pre_tokenizer = pre_tokenizers.Whitespace()

# Example sentence
text = "unhappiness"

# Training the tokenizer with a sample corpus
trainer = trainers.BpeTrainer(vocab_size=1000)
tokenizer.train_from_iterator([text], trainer=trainer)

# Encode the text
encoded = tokenizer.encode(text)
print("Subword Tokenization (BPE):", encoded.tokens)
```

```python
from tokenizers import Tokenizer, models, pre_tokenizers, trainers

# Create a BPE tokenizer
tokenizer = Tokenizer(models.BPE())
tokenizer.pre_tokenizer = pre_tokenizers.Whitespace()

# Sample training corpus (long text)
corpus = [
    "Happiness and sadness are part of life.",
    "Unhappiness does not last forever.",
    "People experience different emotions throughout their lives.",
    "Joy, sorrow, excitement, and fear shape our experiences.",
    "Emotional intelligence helps us understand our own feelings.",
    "It is important to stay positive even during tough times.",
    "Hardships teach us valuable life lessons.",
    "Every challenge brings new opportunities.",
    "A positive mindset can transform difficulties into growth.",
    "Understanding emotions allows us to connect with others deeply."
]

# Train the BPE tokenizer
trainer = trainers.BpeTrainer(vocab_size=1000)
tokenizer.train_from_iterator(corpus, trainer=trainer)

# Encode a test word
text = "unhappiness"
encoded = tokenizer.encode(text)
print("Subword Tokenization (BPE):", encoded.tokens)
```

5️⃣ Hybrid Tokenization (Subword + Word Tokenization, used in GPT, BERT, T5)
Using Hugging Face’s Transformers

```python
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")

text = "Transformers are amazing!"
tokens = tokenizer.tokenize(text)
print("Hybrid Tokenization (BERT-style):", tokens)
```

```python
import faiss
import numpy as np

# Create a vector database with 5D embeddings
d = 5  # Vector dimension
index = faiss.IndexFlatL2(d)  # L2 distance (Euclidean)

# Sample word embeddings (5-dimensional)
word_embeddings = np.array([
    [0.2, 0.8, -0.5, 0.6, 1.0],  # King
    [0.3, 0.7, -0.4, 0.5, 0.9],  # Queen
    [0.9, 0.2, 0.7, -0.3, 0.1]   # Apple
], dtype=np.float32)

# Add to FAISS index
index.add(word_embeddings)

# Query: Find similar words to "King"
query_vector = np.array([[0.2, 0.8, -0.5, 0.6, 1.0]], dtype=np.float32)
D, I = index.search(query_vector, k=2)  # Find top 2 similar words
print(I)
```
